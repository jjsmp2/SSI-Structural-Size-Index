```python
import pandas as pd
import re

# Load dataset
df = pd.read_csv("/kaggle/input/datasets/jigishadas/loccyclomaticcomplexitycalculation/codesearchnet_sample.csv")


# Function: Lines of Code (LOC)

def compute_loc(code):
    lines = code.split("\n")

    # Remove blank lines
    lines = [l for l in lines if l.strip() != ""]

    return len(lines)



# Function: McCabe Cyclomatic Complexity

def compute_cyclomatic(code):

    complexity = 1

    patterns = [
        r"\bif\b",
        r"\belse\s+if\b",
        r"\belif\b",
        r"\bfor\b",
        r"\bwhile\b",
        r"\bcase\b",
        r"\bcatch\b",
        r"\bexcept\b",
        r"\?\s*",      # ternary operator
        r"&&",
        r"\|\|"
    ]

    for p in patterns:
        complexity += len(re.findall(p, code))

    return complexity


# Compute metrics
df["LOC"] = df["whole_func_string"].apply(compute_loc)
df["Cyclomatic_Complexity"] = df["whole_func_string"].apply(compute_cyclomatic)

# Keep only required columns
metrics = df[[
    "language",
    "repository_name",
    "func_name",
    "LOC",
    "Cyclomatic_Complexity"
]]

metrics.to_csv("traditional_metrics.csv", index=False)

print(metrics.head())
```
